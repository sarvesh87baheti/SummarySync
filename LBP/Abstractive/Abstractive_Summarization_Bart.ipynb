```python

# Import necessary libraries
from transformers import pipeline

# Initialize the summarization pipeline
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")
```

```python
# Take user input for the text to summarize
text = input("Enter the text you want to summarize: ")

```

```python

# Generate the summary
summary = summarizer(text, do_sample=False)

# Display the original text and the summary
print("\nOriginal Text:\n", text)
print("\nGenerated Summary:\n", summary[0]['summary_text'])
```
